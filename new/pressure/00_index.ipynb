# Team pressure analysis - index (v32, split)
The 87-cell notebook is split into seven that run in order. Shared setup is in `pressure_common.py` (same folder).

| # | notebook | what it covers |
|---|---|---|
| 1 | 01_pressure_signal.ipynb | B. The pressure signal |
| 2 | 02_where_zone_channel_line.ipynb | C. WHERE - zone, channel, defensive line, shape |
| 3 | 03_timing_and_possession.ipynb | D. WHEN - reaction speed and possession-level pressure |
| 4 | 04_passes_outcomes_triggers.ipynb | E/F. EFFECT and TRIGGERS |
| 5 | 05_summary_and_registry.ipynb | G/H. Summary + event registry |
| 6 | 06_clips.ipynb | H. Clips |
| 7 | 07_exports.ipynb | I. Exports |

**Data**: reads `player_frame_table_named.parquet`, `ball_frame_table_named.parquet` and `annotated_teams.mp4` from `new_cache`; writes everything to `new_cache/pressure_analysis/` (`state/` hand-off pickles, `clips/`, `stoppage_check/`, the two json exports, `clip_index.csv`).

**Typical use**: run 01 -> 05 once (a few minutes); then 06 (clips, the slow one) and 07 (exports). 06 can be re-run alone with different settings, as long as 01-05 have run.

**One setting to know**: `FLIP_CHANNELS` in `pressure_common.py` (left / right mirrored vs the video).

### Optional: run all seven from here
Needs `jupyter nbconvert` in the same environment. Each notebook is executed in place (outputs are saved into the notebook).

```
import subprocess, sys, time
NOTEBOOKS = ["01_pressure_signal", "02_where_zone_channel_line", "03_timing_and_possession", "04_passes_outcomes_triggers",
             "05_summary_and_registry", "06_clips", "07_exports"]
for nbname in NOTEBOOKS:                      # remove items from the list to run only some
    t0 = time.time()
    r = subprocess.run([sys.executable, "-m", "jupyter", "nbconvert", "--to", "notebook", "--execute", "--inplace",
                        "--ExecutePreprocessor.timeout=-1", f"{nbname}.ipynb"], capture_output=True, text=True)
    print(f"{nbname}: {'ok' if r.returncode == 0 else 'FAILED'} ({time.time() - t0:.0f}s)")
    if r.returncode != 0:
        print(r.stderr[-1500:]); break
```
